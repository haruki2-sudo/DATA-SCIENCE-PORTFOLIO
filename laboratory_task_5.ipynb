**1. Perform Standard Imports**

```python
import torch
import numpy as np
```

**2. Create a function called `set_seed()` that accepts `seed: int` as a parameter, this function must return nothing but just set the seed to a certain value.**

```python
def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
```

**3. Create a NumPy array called "arr" that contains 6 random integers between 0 (inclusive) and 5 (exclusive), call the `set_seed()` function and use `42` as the seed parameter.**

```python
set_seed(42)
arr = np.random.randint(0, 5, size=6).astype('int32')

assert np.array_equal(arr, np.array([3,4,2,4,4,1])), "Oops"
print(arr)
```

**4. Create a tensor "x" from the array above**

```python
x = torch.from_numpy(arr)
print(x)
```

**5. Change the dtype of x from `int32` to `int64`**

```python
print('Old:', x.type())

x = x.type(torch.int64)

print('New:', x.type())
```

**6. Reshape `x` into a 3x2 tensor**

There are several ways to do this.

```python
x = x.reshape(3,2)
print(x)
```

**7. Return the right-hand column of tensor `x`**

```python
print(x[:,1:])
```

**8. Without changing x, return a tensor of square values of `x`**

There are several ways to do this.

```python
# x.pow(2) returns a NEW tensor of squared values without modifying x in place
# (printing the result explicitly, since the original scaffold's `x.pow(2)` alone
# on its own line doesn't display anything when run as a .py-style script cell)
print(x.pow(2))

# Confirm x itself is unchanged
print(x)
```

**9. Create a tensor `y` with the same number of elements as `x`, that can be matrix-multiplied with `x`**

Use PyTorch directly (not NumPy) to create a tensor of random integers between 0 (inclusive) and 5 (exclusive). Use 42 as seed.

Think about what shape it should have to permit matrix multiplication.

```python
set_seed(42)
y = torch.randint(0,5,(2,3))

assert torch.equal(y, torch.tensor([[2, 2, 1],[4, 1, 0]])), "Oops"
print(y)
```

**10. Find the matrix product of `x` and `y`.**

```python
prod = torch.mm(x,y)
prod
```

## Summary

| Step | Result |
|---|---|
| `arr` | `[3, 4, 2, 4, 4, 1]` |
| `x` (after reshape, dtype int64) | `[[3, 4], [2, 4], [4, 1]]` |
| Right-hand column of `x` | `[[4], [4], [1]]` |
| `x.pow(2)` (x itself unchanged) | `[[9, 16], [4, 16], [16, 1]]` |
| `y` | `[[2, 2, 1], [4, 1, 0]]` |
| `x @ y` (3x2 · 2x3 → 3x3) | shown in the final cell above |

`x` is a 3×2 tensor and `y` is a 2×3 tensor, so `torch.mm(x, y)` is valid (inner dimensions
match: 2 = 2) and produces a 3×3 result — this is why `y` had to be shaped `(2, 3)` in step 9
rather than `(3, 2)` or any other shape.
